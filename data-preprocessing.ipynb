# Data Preprocessing

1. Renaming the columns
2. Dropping email and timestamp
3. Changing the value of data into numerical for statistical work

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('response.csv')
df.head()
```

```python
# 1. Renaming the columns
df.columns = [
    'timestamp', 
    'email',
    'membership_status', 
    'age_group', 
    'visit_frequency',
    'notice_cfp_label',
    'motivated_by_eco_points',
    'pos_convenience',
    'preferred_eco_benefits',
    'feeling_on_digital_receipt'
]
df.head()
```

```python
# 2. Dropping email and timestamp
df = df.drop(columns=['timestamp', 'email'])
df.head()
```

```python
# 3. Changing the value of data into numerical for statistical work

mem_map = {
    'เป็นสมาชิก M-Card': 1,
    'ยังไม่ได้เป็น': 0
}
df['membership_status'] = df['membership_status'].map(mem_map)

freq_map = {
    'สัปดาห์ละหลายครั้ง': 3,
    'สัปดาห์ละครั้ง': 3,
    'เดือนละ 1-2 ครั้ง': 2,
    'นานๆครั้ง': 1
}
df['visit_frequency'] = df['visit_frequency'].map(freq_map)

notice_map = {
    'ประจำ': 3,
    'บางครั้ง (ถ้ามองเห็นชัดเจน)': 2,
    'ไม่เคยสังเกตเลย': 1
}
df['notice_cfp_label'] = df['notice_cfp_label'].map(notice_map)

motiv_map = {
    'มากที่สุด (พร้อมเปลี่ยนยี่ห้อทันที)': 3,
    'ปานกลาง (พิจารณาประกอบกับราคาและคุณภาพ)': 2,
    'น้อย / ไม่มีผลต่อการตัดสินใจ': 1
}
df['motivated_by_eco_points'] = df['motivated_by_eco_points'].map(motiv_map)

conv_map = {
    'สะดวกมาก (ชอบที่ไม่ต้องทำขั้นตอนเพิ่ม)': 3,
    'ปานกลาง': 2,
    'ไม่สะดวก': 1
}
df['pos_convenience'] = df['pos_convenience'].map(conv_map)

feel_map = {
    'ช่วยให้ตระหนักเรื่องสิ่งแวดล้อมและอยากซื้อสินค้ารักษ์โลกเพิ่มขึ้น': 3,
    'เฉยๆ / ไม่ได้สนใจข้อมูลนี้': 1
}
df['feeling_on_digital_receipt'] = df['feeling_on_digital_receipt'].map(feel_map)

# One-hot encode preferred_eco_benefits and rename columns
benefits_dummies = df['preferred_eco_benefits'].str.get_dummies(sep=', ')
benefits_dummies = benefits_dummies.rename(columns={
    'นำคะแนนไปร่วมบริจาคโครงการสิ่งแวดล้อม': 'donate_to_environment',
    'ส่วนลดคูปองช้อปปิ้ง/คูปองเงินสดในห้างเดอะมอลล์': 'shopping_discount_coupon',
    'แลกรับสินค้ารักษ์โลกพิเศษ': 'redeem_eco_product'
})
df = pd.concat([df.drop(columns=['preferred_eco_benefits']), benefits_dummies], axis=1)

df.head()
```

```python
# Save the cleaned data
df.to_csv('cleaned_response.csv', index=False)
print('Data preprocessing complete. Saved to cleaned_response.csv')
```
